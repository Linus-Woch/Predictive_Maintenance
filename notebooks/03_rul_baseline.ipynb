```python
import pandas as pd

df_validation = pd.read_csv("../outputs/VALIDATION_FD001.csv")
df_fitting = pd.read_csv("../outputs/FITTING_FD001.csv")
```

As a very first prediction method, we will now use the median RUL among the fitting_df values and check it with the validation set. We will use MAE and RMSE to do a first "performance" review, expecting to have a lot of room for improvement still

```python
median_RUL = df_fitting["RUL"].median()
print(f"Median RUL: {median_RUL}")

df_validation["median_prediction"] = median_RUL
errors = df_validation["median_prediction"] - df_validation["RUL"]

MAE = errors.abs().mean()
RMSE = (errors ** 2).mean() ** 0.5

print(f"Mean Absolute Error: {MAE}")
print(f"Root Mean Squared Error: {RMSE}")
```
